```python
import numpy as np
```

```python

arr=np.random.randint(1,101,size=(3,3))
print(arr)
print("--")
max_val=np.max(arr)
min_val=np.min(arr)
normalized_arr=(arr - min_val ) / (max_val - min_val)
print("normalized_arr")
normalized_arr
```

```python
arr1=np.random.randint(1,20,size=10)

print(arr1)
x=int(input("enter X"))
closest_value = arr1[np.argmin(np.abs(arr1 - x))]

print("Closest value:", closest_value)




```

```python
arr3 = np.random.randint(1, 100, size=(5, 5))
print(arr3)
print("----")
sorted_arr3 = np.sort(arr3, axis=0)
print("sorted_arr3",sorted_arr3)
result=sorted_arr3[3,:]
print("-----")
print(result)
```

```python
arr = np.random.randint(1, 100, size=(6, 4))
print(arr)
row_sum = np.sum(arr, axis=1)
result = np.where(row_sum % 2 == 0)
print(result[0])
```

```python
arr=np.random.randint(1,51,size=30)
print(arr)

arr[(arr % 3==0) |( arr % 5==0)]=-1
print(arr)

```

```python


arr = np.random.randint(1, 100, size=(3, 3))

print(arr)
print("---")

row_mean = np.mean(arr, axis=1)
print(row_mean)

result = arr - row_mean[:, np.newaxis]
print("--")

print(result)
```
